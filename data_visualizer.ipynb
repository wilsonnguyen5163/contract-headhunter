```python
import os
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

```python
contractors = pd.read_csv(r"/home/wilson/Work/do_good/contract-headhunter/data/contractors/contractors.csv")
```

```python
def view_talent_details(talent_id):
    talent = contractors[contractors['talent_id'] == talent_id]
    data = pd.DataFrame([talent['talent_name'], talent['location'], talent['executive_summary'],  talent['main_industry'], talent['core_strengths'], talent['proof_points'],
                         talent['target_titles']])
    return data
```

```python
import glob
from IPython.display import display

def view_trial_run(path=None, band=None):
    """Best 2 contractors per job from a trial-run CSV (latest in exports/ if no path).
    band: optional filter, e.g. 'high' or ['high', 'medium']; 'no match' shows jobs with none."""
    if path is None:
        path = sorted(glob.glob(r"/home/wilson/Work/do_good/contract-headhunter/exports/trial-run-*.csv"))[-1]
    trial = pd.read_csv(path, dtype=str, keep_default_na=False)
    if band is not None:
        bands = [band] if isinstance(band, str) else band
        trial = trial[trial['band'].isin(bands)]
    view = trial[['job_title', 'company', 'option', 'score', 'band', 'talent_name', 'talent_title', 'shared_skills']].copy()
    view.loc[trial['job_url'].duplicated(), ['job_title', 'company']] = ''  # show each job once
    print(os.path.basename(path), '-', trial['job_url'].nunique(), 'jobs')
    with pd.option_context('display.max_rows', None):  # every row, no '...' in the middle
        display(view.reset_index(drop=True))
```

```python
view_trial_run()
```
